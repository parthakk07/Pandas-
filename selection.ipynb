# selection 

```python
import pandas as pd 
```

```python
df=pd.read_csv("data.csv")
print(df)
```

# selection by column

```python
print(df["Name"])
```

```python
print(df["Weight"].to_string())
```

```python
print(df[["Name","Weight","Height"]])
```

# selection with row 

```python
print(df.loc[0])
```

# making name to use as index 

```python
df=pd.read_csv("data.csv",index_col="Name")
print(df)
```

```python
print(df.loc["Pikachu"])
```

```python
print(df.loc["Charizard",["Height","Weight"]])
```

```python
# all after charizard 
print(df.loc["Charizard":])
```

```python
# can also use iloc fo rindex based selection 
print(df.iloc[0:11])
```

```python
# also folloows rows,cloumn 
print(df.iloc[0:11:2,0:3])
```
