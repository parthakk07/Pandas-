```python
import numpy as np
import pandas as pd
import matplotlib as plt 
```

# series -1d array

```python
data=[100,120,140]
series1=pd.Series(data)
print(series1)

series2=pd.Series(data,index=["a","b","c"])
print(series2)

print("a=",series2.loc["a"])
```

```python
series=pd.Series(data,index=["a","b","c"])
print(series)
series.loc["c"]=200
print(series)
```

```python
print(series.iloc[0])
print(series.iloc[1])
print(series.iloc[2])
```

```python
data1=[100,120,140,160,180,200,220]
series=pd.Series(data1,index=["a","b","c","d","e","f","g"])
print(series)
print("condition")
print(series[series >=200])
```

```python
steps={"day 1":1640,"day 2":1423,"day 3":1088 }
series=pd.Series(steps)
series.loc["day 3"] +=20
print(series)
```

## dataframes -2d

```python
data={"name": ["nirav","parth","brocode"],
      "age":[13,18,28]}

df=pd.DataFrame(data,index=["employee 1","employee 2" ,"employee 3"])
print(df)
print(df.loc["employee 2"])
print(df.iloc[1])


#add column
df["language"] =["rust","python","java"]
print(df)


# add new row
new_row=pd.DataFrame({"name":"harry","age":30,"language":"binary"},index=["employee 4"])
df=pd.concat([df,new_row])
print(df)
```
