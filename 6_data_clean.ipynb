data cleaning -the process of fixing /removing incomplete ,incorrect data 

```python
import pandas as pd
```

```python
df=pd.read_csv("fkedata.csv")
print(df)
```

# drop column --> irrelevant column 

```python
df=df.drop(columns=["Legendary","No"])
print(df)
```

#  handlinh missing data

```python
df=df.dropna(subset=["Type2"])
print(df)
```

```python
df=df.fillna({"Type2": "none"})
print(df)
```

```python
df["Type1"]=df["Type1"].replace({"Grass":"GRASS",
                                 "Fire":"FIRE"})
print(df)
```

# standardize text

```python
df["Name"]=df["Name"].str.lower()
print(df)
```

# fix data type

```python
df["Legendary"]=df["Legendary"].astype(bool)
print(df)
```

# fix multiple data

```python
df=df.drop_duplicates()
print(df)
```

```python
df=df.isnull().sum()
print(df)
```

```python
df["Height"]=df["Height"].apply(lambda x: x*2)
print(df)
```
